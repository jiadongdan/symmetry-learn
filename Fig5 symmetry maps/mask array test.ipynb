```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
import numpy.ma as ma
```

```python
# Create a 2D numpy array
data = np.random.rand(10, 10)  # Example data

# Create a mask (e.g., masking the top-left 3x3 region)
mask = np.zeros_like(data, dtype=bool)
mask[:3, :3] = True  # Masking the top-left 3x3 region

# Apply the mask to the data
masked_data = ma.array(data, mask=mask)

# Plot using imshow
plt.imshow(masked_data, cmap='viridis', interpolation='none')
```
