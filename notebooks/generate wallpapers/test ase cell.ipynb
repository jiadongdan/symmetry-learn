```python
from ase import Atoms

# Example: Create an Atoms object with a larger unit cell
atoms = Atoms('C2', positions=[[0, 0, 0], [8, 8, 8]], cell=[10, 10, 10])

pos1 = atoms.get_scaled_positions()

# Shrinking the unit cell without rescaling atomic positions (default behavior)
atoms.set_cell([5, 5, 5], scale_atoms=False)  # Atoms remain at original positions

pos2 = atoms.get_scaled_positions()
```
