```python
from mtflearn import ZPs

%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt 
```

```python
def compute_kernels(n_max, size):
    zps = ZPs(n_max=n_max, size=size)
    kernels = zps.polynomials
    mask1 = zps.m > 1
    mask2 = zps.m < -1
    print(zps.m[zps.m>1])
    print(np.abs(zps.m[zps.m<-1]))
    A_kernels = kernels[mask1]
    B_kernels = kernels[mask2]
    return A_kernels, B_kernels

```

```python
aa, bb = compute_kernels(n_max=10, size=45)
```
